```python
#Imports for opensmile
import librosa
import numpy as np
import pandas as pd
import os
import opensmile
print(np.__version__) 
import typing_extensions as tp
print(tp.__version__) 
```

```python
#go  through folders in "study_data_clean" an, one by one, read only  the   mp3 files, make compatible with opensmile, and extract features

#path to the folder with the mp3 files
path = "./study_data_clean/study_sessions"
folders = os.listdir(path)
print(folders)
for session in folders:
    path_to_session = path + "/" + session
    files = os.listdir(path_to_session)
    for file in files:
        #read only mp3 files 
        if file.endswith(".mp3"):
            #read the file
            y, sr = librosa.load(path_to_session + "/" + file, sr=44100)
            ##convert to mono
            #y = librosa.to_mono(y)
            ##convert to 16bit
            #y = librosa.util.fix_length(y, 132300)
            #extract features
            smile = opensmile.Smile(
                feature_set=opensmile.FeatureSet.ComParE_2016,
                feature_level=opensmile.FeatureLevel.LowLevelDescriptors,
            )
            results = smile.process_signal(y, sr)
            #save the features to a csv file
            results.to_csv("./opensmile_feats/" + session + "_" + file[:-4] + ".csv")

            
```

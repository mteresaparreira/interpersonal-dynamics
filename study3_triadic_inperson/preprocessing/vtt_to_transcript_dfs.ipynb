```python
#imports
import numpy as np
import pandas as pd
import os
import json
```

```python
#make new folder, called "transcript_dfs"
os.system('mkdir ../2024_study3_data/transcript_dfs')
```

```python
path = '../2024_study3_data/extracted_audio_transcripts/'



listfiles = os.listdir(path)
listfiles = [x for x in listfiles if x[-4:] == '.vtt']

for file in listfiles:
    #open vtt file
    with open(path+file) as f:
        lines = f.readlines()
    #delete first line
    lines = lines[1:]
    #delete lines that are empty ('\n)
    lines = [x for x in lines if x != '\n']

    #start dataframe
    df = pd.DataFrame(columns=['start','end','speaker','text'])

    #loop through lines
    for line in lines:
        #remove '\n' from end of line
        line = line[:-1]
        #if line starts with a number, it is a timestamp
        if line[0].isdigit():
            #split line into start and end time
            start = line.split(' --> ')[0]
            end = line.split(' --> ')[1]
            #print(start,end)
        #if line starts with a letter, it is a speaker
        elif line[0].isalpha():
            #split line into speaker and text
            speaker = line.split(':')[0]
            text = line.split(':')[1]
            #print(speaker,text)
            #add row to dataframe
            row_df = pd.DataFrame({'start':[start],'end':[end],'speaker':[speaker],'text':[text]})
            df = pd.concat([df,row_df])

    #reset index
    df.reset_index(drop=True,inplace=True)

    #save dataframe as csv
    df.to_csv('../2024_study3_data/transcript_dfs/'+file[:-4]+'.csv',index=False)
```

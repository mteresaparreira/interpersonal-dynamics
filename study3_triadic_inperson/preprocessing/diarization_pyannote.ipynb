```python
from pyannote.audio import Pipeline
import torch
import os
import matplotlib.pyplot as plt

# Initialize the pipeline
pipeline = Pipeline.from_pretrained(
    "pyannote/speaker-diarization-3.1",
    use_auth_token="REDACTED_HF_TOKEN"
)

class format:
   BOLD = '\033[1m'
   END = '\033[0m'


pipeline.to(torch.device("cuda"))

diarizations = []

# Iterate through files 1-34
for i in range(1, 35):
    file_path = f"data/{i}.wav"

    if os.path.exists(file_path):
        # if not os.path.exists(f"output/{i}.rttm"):
        with ProgressHook() as hook:
            diarization = pipeline(file_path, num_speakers=3, hook=hook)

        # Dump the diarization output to disk using RTTM format
        with open(f"output/{i}.rttm", "w") as rttm:
            diarization.write_rttm(rttm)

        display(diarization)
        print(format.BOLD + f'Visualization of {file_path}' + format.END)
    else:
        print(f"File {file_path} does not exist.")
```
